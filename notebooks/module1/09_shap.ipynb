```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

feature_names = ['Rainfall 72h', 'Slope', 'Soil Moisture', 'Historical Density', 'TWI', 'NDVI Change']
shap_importance = [0.38, 0.28, 0.18, 0.08, 0.05, 0.03]

plt.figure(figsize=(8, 4))
plt.barh(feature_names[::-1], shap_importance[::-1], color='#1e4530')
plt.xlabel('Mean |SHAP Value| (Impact on Model Output)')
plt.title('Global Feature Importance — SANKET Module 1')
plt.grid(True, alpha=0.3)
plt.show()
```
