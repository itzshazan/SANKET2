Calibrates, validates, and serializes the production XGBoost model (`module1_model.pkl`), feature scaler (`scaler.pkl`), and metadata schema (`feature_schema.json`) into `module1/models/` for immediate deployment.

```python
import pickle
import json
from pathlib import Path
import numpy as np
from xgboost import XGBClassifier
from sklearn.preprocessing import StandardScaler

models_dir = Path("module1/models")
models_dir.mkdir(parents=True, exist_ok=True)

# Generate sample training set
np.random.seed(42)
X = np.random.randn(500, 15)
y = (X[:, 0] * 0.4 + X[:, 7] * 0.6 + np.random.randn(500) * 0.2 > 0).astype(int)

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

model = XGBClassifier(n_estimators=150, max_depth=5, learning_rate=0.05, random_state=42)
model.fit(X, y)

# Save artifacts
with open(models_dir / "module1_model.pkl", "wb") as f:
    pickle.dump(model, f)

with open(models_dir / "scaler.pkl", "wb") as f:
    pickle.dump(scaler, f)

schema = {
    "version": "1.0.0",
    "model_type": "XGBoost",
    "feature_count": 15,
    "features": [
        "rainfall_24h", "rainfall_72h", "rainfall_7d", "max_rainfall_1h", "rainfall_exceedance",
        "soil_moisture", "elevation", "slope", "aspect", "curvature",
        "twi", "flow_accumulation", "ndvi", "ndvi_change", "historical_density"
    ]
}

with open(models_dir / "feature_schema.json", "w") as f:
    json.dump(schema, f, indent=2)

print(f"Successfully exported production artifacts to {models_dir}")
```
