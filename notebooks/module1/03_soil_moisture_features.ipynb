Extracts and calibrates surface soil Volumetric Water Content (VWC, 0-5cm) from NASA SMAP satellite radiometry and computes saturation ratios.

```python
import numpy as np
import matplotlib.pyplot as plt

def compute_soil_saturation(vwc_m3_m3, porosity=0.50):
    saturation_ratio = np.clip(vwc_m3_m3 / porosity, 0.0, 1.0)
    if saturation_ratio >= 0.85:
        status = 'SATURATED'
    elif saturation_ratio >= 0.65:
        status = 'ELEVATED'
    elif saturation_ratio >= 0.45:
        status = 'NORMAL'
    else:
        status = 'LOW'
    return saturation_ratio, status

test_vwc = [0.15, 0.28, 0.38, 0.48]
for v in test_vwc:
    sr, status = compute_soil_saturation(v)
    print(f"VWC: {v:.2f} m3/m3 -> Saturation: {sr*100:.1f}% ({status})")
```
