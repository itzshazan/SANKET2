```python
import pandas as pd
import numpy as np

np.random.seed(42)
n_samples = 1000

df = pd.DataFrame({
    'rainfall_24h': np.random.exponential(35, n_samples),
    'rainfall_72h': np.random.exponential(80, n_samples),
    'soil_moisture': np.random.uniform(0.15, 0.52, n_samples),
    'slope': np.random.uniform(5, 60, n_samples),
    'elevation': np.random.uniform(200, 3200, n_samples),
    'twi': np.random.uniform(4, 16, n_samples),
    'ndvi': np.random.uniform(0.2, 0.85, n_samples),
    'ndvi_change': np.random.normal(-0.02, 0.08, n_samples),
    'historical_density': np.random.poisson(2, n_samples),
})

# Physics-based landslide trigger probability
risk_score = (
    0.02 * df['rainfall_72h'] +
    0.06 * df['slope'] +
    4.5 * df['soil_moisture'] +
    0.3 * df['historical_density'] - 3.8
)
prob = 1.0 / (1.0 + np.exp(-risk_score))
df['target'] = (prob > 0.45).astype(int)

print("Class distribution:
", df['target'].value_counts())
df.to_parquet('data/final/dataset.parquet', index=False)
print("Saved to data/final/dataset.parquet")
```
