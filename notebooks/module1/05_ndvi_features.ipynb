```python
import numpy as np

def calculate_ndvi(red_band, nir_band):
    return (nir_band - red_band) / (nir_band + red_band + 1e-8)

red = np.array([0.08, 0.12, 0.25])
nir = np.array([0.55, 0.45, 0.28])
ndvi = calculate_ndvi(red, nir)
for i, n in enumerate(ndvi):
    print(f"Sample {i+1}: NDVI = {n:.3f}")
```
