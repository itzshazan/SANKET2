Extracts terrain slope, aspect, plan/profile curvature, and Topographic Wetness Index (TWI) from Copernicus DEM 30m.

```python
import numpy as np

def compute_twi(flow_accumulation, slope_degrees):
    tan_slope = np.tan(np.radians(max(slope_degrees, 1.0)))
    return np.log(flow_accumulation / tan_slope)

slopes = [10, 25, 38, 55]
for s in slopes:
    twi = compute_twi(150.0, s)
    print(f"Slope: {s}° -> TWI: {twi:.2f}")
```
