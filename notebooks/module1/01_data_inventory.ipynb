**Goal**: Inventory, verify, and document data feeds for Landslide Risk Prediction in the North Eastern Region (NER).

| Source | Variables | Format | Frequency |
|---|---|---|---|
| **ISRO Landslide Atlas** | Historical failure events | CSV / GeoJSON | Static Catalog |
| **NASA GPM IMERG** | Precipitation accumulation (24h/72h/7d) | HDF5 / NetCDF4 | 30-minute |
| **NASA SMAP** | Volumetric Water Content (VWC) | HDF5 | Daily |
| **Copernicus DEM** | Elevation, Slope, Aspect, TWI | GeoTIFF 30m | Static Raster |
| **Sentinel-2 L2A** | B04 (Red), B08 (NIR) for NDVI | SAFE / Cloud GeoTIFF | 5-day revisit |
