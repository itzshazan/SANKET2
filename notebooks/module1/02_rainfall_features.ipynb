# SANKET Module 1 — Notebook 02: Rainfall Feature Engineering

Derives rolling rainfall accumulation (24h, 72h, 7d), Antecedent Precipitation Index (API), and regional Intensity-Duration threshold exceedance $E = I / I_0(D)$.

Regional empirical power law for North East India:
$$I_0(D) = 5.8294 \cdot D^{-0.4141}$$

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

def compute_threshold_intensity(duration_hours):
    return 5.8294 * (duration_hours ** -0.4141)

durations = np.linspace(1, 168, 100) # 1 hour to 7 days
threshold_curve = compute_threshold_intensity(durations)

plt.figure(figsize=(8, 4.5))
plt.plot(durations, threshold_curve, 'r-', lw=2, label='NE Himalaya Threshold $I_0(D)$')
plt.xlabel('Storm Duration (hours)')
plt.ylabel('Threshold Intensity (mm/h)')
plt.title('Calibrated Rainfall Threshold for NER')
plt.grid(True, alpha=0.3)
plt.legend()
plt.show()
```

```python
def compute_rainfall_features(rainfall_24h, rainfall_72h, rainfall_7d):
    duration = 72.0
    intensity = rainfall_72h / duration
    i_0 = compute_threshold_intensity(duration)
    exceedance = intensity / i_0
    api = rainfall_24h + (rainfall_72h - rainfall_24h) * 0.90
    return {
        'rainfall_24h': rainfall_24h,
        'rainfall_72h': rainfall_72h,
        'rainfall_7d': rainfall_7d,
        'rainfall_exceedance': round(exceedance, 3),
        'api_index': round(api, 2)
    }

sample_rf = compute_rainfall_features(134.0, 280.0, 395.0)
print("Tupul Storm Features:", sample_rf)
```
